```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from openai import OpenAI
from docx import Document
from docx.enum.text import WD_ALIGN_PARAGRAPH
data=[]
def post(i):
    url="https://search.ltn.com.tw/list?keyword=%E6%88%BF%E5%9C%B0%E7%94%A2&&sort=date&type=all&page={}".format(i)
    req=requests.get(url)
    soup=BeautifulSoup(req.text,"html.parser")
    for i in soup.find_all("div",class_="cont"):
        page=i.find("p").text
        data.append({"內容":page})
    print(data)
    return pd.DataFrame(data)
df=[]
for p in range(1,2):
    w=post(p)
    df.append(w)
f=pd.concat(df,ignore_index=True)
f.to_excel("d:\新聞類\POST.xlsx")
```

```python
df=pd.read_excel(r"d:\新聞類\POST.xlsx")
df1="".join(df["內容"].fillna("").astype(str))
max_rows = 30  # 限制只抓前 50 筆新聞來分析（可依你的 Token 額度自行調整）
df_subset = df.head(max_rows)

# 將選定的新聞內容合併，並適度控制長度
df1 = "\n".join(df_subset["內容"])

# 如果文本還是太長，可以強制切斷（例如限制最多 8000 個字）
if len(df1) > 8000:
    df1 = df1[:8000]
    print("提示：新聞文本過長，已自動截斷前 8000 字以避免超出 API 限制。")
client=OpenAI(
    api_key="",
    base_url="https://api.groq.com/openai/v1",
)
def api_ai(user_input_text):
    pmotpm=f"""
    請嚴格依照以下結構與專業金融語氣進行輸出，確保內容具備商業價值，適合呈報給金控公司、銀行授信部門或資產管理高層：

1. 宏觀政策與金融風向
   - 涵蓋項目：央行利率動態、房市調控政策、稅務變動及對市場的具體影響。
   - 格式要求：使用表格呈現（欄位包含：事件／政策、主要訊息、對市場的可能影響）。

2. 市場供需與價格趨勢
   - 涵蓋項目：高價住宅（豪宅）佔比、重點區域租金行情、新建案價格波動、法拍市場與供給量變化。
   - 附帶市場心理分析（買方與賣方心態）。

3. 熱點區域與產品焦點
   - 涵蓋項目：核心精華區（如北市信義區高價宅）、重劃區／科技園區新案、低價或法拍潛力區。
   - 提供具體的投資與購屋吸引力評估。

4. 潛在風險與警訊
   - 涵蓋項目：外部經濟衝擊、供給過剩風險、法拍物件風險、政策打壓與資金流向危機。

5. 首購與剛需市場（低於 3 千萬預算區段）
   - 涵蓋項目：熱門行政區推薦（如北投、士林、三重、林口等）、平均售價與生活機能配套建議。

6. 總結與核心建議
   - 綜合以上數據，給出明確的市場趨勢總結與投資／授信風險建議。

要求：
- 語氣必須客觀、專業、精準，多使用量化數據與結構化表格。
- 嚴格確保時間跨度與內容為「月度（Month）」視野，切勿寫成單週內容。

---

    {user_input_text}
    """
    responst=client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {"role":"system","content":"你是一位資深的房地產市場分析師與金融風控研究員。請幫我撰寫一份專業的「台灣房地產市場與總體經濟月度分析報告」（時間範圍涵蓋：[請填入年月，例如 2026年8月]）。"},
            {"role":"user","content": pmotpm}
        ],
        temperature=0.3
    )
    return responst.choices[0].message.content
pf=api_ai(df1)
print(pf)
```

```python
doc=Document()
a=doc.add_heading("房地產戰報",level=1)
a.alignment=WD_ALIGN_PARAGRAPH.CENTER
doc.add_heading("房地產市場週報與趨勢分析",level=1)
doc.add_paragraph(pf)
doc.save("d:\新聞類\房地產資料.docx")
```
